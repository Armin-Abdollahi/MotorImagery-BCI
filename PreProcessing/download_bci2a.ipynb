```python
"""
Step 1: Download and verify raw BCI Competition IV-2a (BNCI2014_001)
Does not perform any filtering, epoching, or feature manipulation.
"""

import sys

try:
    from moabb.datasets import BNCI2014_001
except ImportError:
    try:
        from moabb.datasets import BNCI2014001 as BNCI2014_001
    except ImportError:
        print("[!] MOABB is not installed. Please install via: pip install moabb")
        sys.exit(1)


def download_and_verify():
    print(">>> Initializing BNCI2014_001 (BCI Competition IV-2a)...")
    dataset = BNCI2014_001()
    subjects = list(range(1, 10))  # 9 subjects: 1 to 9

    print(f">>> Target subjects: {subjects}")
    print(">>> Starting automated download (files will be cached locally)...")

    file_paths = {}
    for sub in subjects:
        print(f"[*] Downloading / Checking Subject {sub}...")
        # data_path automatically downloads if not present on disk
        paths = dataset.data_path(sub)
        file_paths[sub] = paths
        print(f"    -> Subject {sub} ready. Files: {len(paths)}")

    print("\n" + "=" * 50)
    print(">>> Verification Summary:")
    all_ready = True
    for sub, paths in file_paths.items():
        print(f"Subject {sub}: {len(paths)} file(s) verified.")
        if len(paths) == 0:
            all_ready = False

    if all_ready:
        print(">>> SUCCESS: All 9 subjects are downloaded and verified!")
    else:
        print(">>> WARNING: Some subject files were not fetched completely.")
    print("=" * 50)


if __name__ == "__main__":
    download_and_verify()
```
